```python
mydataset = {
'Employee_ID': [101, 102, 103, 104, 105],
'Name': ["Alice", "Bob", "Charlie", "Diana", "Edward"],
'Department': ["HR", "IT", "IT", "Marketing", "Sales"],
'Age': [29, 34, 41, 28, 38],
'Salary': [50000, 70000, 65000, 55000, 60000],
'Years_of_Experience': [4, 8, 10, 3, 12],
'Joining_Date': ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
'Gender': ["Female", "Male", "Male", "Female", "Male"],
'Bonus': [5000, 7000, 6000, 4500, 5000],
'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}
dd = pd.DataFrame(mydataset)
print(dd)
```

```python
import numpy as np
import pandas as pd

print("Shape:", dd.shape)

dd.info()

print(dd.describe())

print("First 5 rows:\n", dd.head(5))
print("\nLast 3 rows:\n", dd.tail(3))

print("Average Salary:", dd["Salary"].mean())
print("Total Bonus:", dd["Bonus"].sum())
print("Youngest Employee Age:", dd["Age"].min())
print("Highest Rating:", dd["Rating"].max())

sorted_dd = dd.sort_values(by="Salary", ascending=False)
print(sorted_dd)

def categorize_rating(rating):
  if rating >= 4.5:
    return "Excellent"
  elif rating >= 4.0:
    return "Good"
  else:
    return "Average"


dd["Performance_Category"] = dd["Rating"].apply(categorize_rating)

print("Missing values in each column:\n", dd.isnull().sum())

dd.rename(columns={"Employee_ID": "ID"}, inplace=True)

exp_gt_5 = dd[dd["Years_of_Experience"] > 5]
print("More than 5 years experience:\n", exp_gt_5)

it_dept = dd[dd["Department"] == "IT"]
print("IT Department employees:\n", it_dept)

dd["Tax"] = dd["Salary"] * 0.10

dd.to_csv("modified_employees.csv", index=False)
```
